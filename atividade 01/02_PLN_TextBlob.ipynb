# Exercício 1

Leia o dataset de review do IMDb no link abaixo, e calcule a taxa de acerto entre as avaliações preditas pelo TextBlob, e os rótulos da base de dados.

````
import pandas as pd

resenhas = pd.read_csv("https://github.com/LawrenceDuan/IMDb-Review-Analysis/blob/master/IMDb_Reviews.csv?raw=true")[0:100]
print(resenhas.shape)
resenhas.head()
````

````
def classificar_sentimento(texto):
    analise = TextBlob(texto)
    # Se a polaridade for maior que 0, consideramos a resenha positiva (1)
    # Caso contrário, consideramos negativa (0)
    if analise.sentiment.polarity > 0:
        return 1
    else:
        return 0

resenhas['predicao'] = resenhas['review'].apply(classificar_sentimento)

# 3. Calculando a taxa de acerto comparando as predições com a coluna 'sentiment' original
taxa_acerto = (resenhas['predicao'] == resenhas['sentiment']).mean()

# 4. Imprimindo o resultado formatado
print(f"Taxa de acerto (Acurácia) do TextBlob: {taxa_acerto * 100:.2f}%")

# Opcional: Visualizar como ficou o dataframe com a nova coluna
resenhas[['review', 'sentiment', 'predicao']].head()
````
